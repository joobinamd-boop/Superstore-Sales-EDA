UNIVARIATE ANALYSIS

 ORDERS BY CATEGORY

```python
plt.figure(figsize=(8,5))
sns.countplot(df['Category'])
plt.title('num of orders by Category')
plt.xlabel('Category')
plt.ylabel('orders')
plt.show()
```

                        ORDERS BY CATEGORY
In this chart Office Supplies has the highest number of orders among the other categories.

OREDRS BY REGION

```python
plt.figure(figsize=(8,5))
sns.countplot(df['Region'])
plt.title('orders by region')
plt.xlabel('Region')
plt.ylabel('Orders')
plt.show()
```

                       OREDRS BY REGION
In this bar shows West and East regions are the highest orders in the sales.

SALES BY CATEGORY

```python
category_sales=df.groupby('Category')['Sales'].sum()
category_sales.plot(kind='bar',figsize=(8,5))
plt.title('Total sales by category')
plt.xlabel('Category')
plt.ylabel('Total sales')
plt.xticks(rotation=0)
plt.show()
```

                      SALES BY CATEGORIES  
In this chart shows Technology is the highest total sales among the other categories.

SALES BY REGION

```python
region_sales=df.groupby('Region')['Sales'].sum()
region_sales.plot(kind='bar',figsize=(8,5))
plt.title('Total sales by region')
plt.xlabel('Region')
plt.ylabel('Total sales')
plt.xticks(rotation=0)
plt.show()
```

                        SALES BY RELIGION
This chart shows which is the strogest region in sales. And the strongest region is West and the second one is East.

SALES OVER TIME

```python
sales_over_time=df.groupby(df['Order Date'])['Sales'].sum()
sales_over_time.plot(kind='line',figsize=(12,5))
plt.title('Total sales over time')
plt.xlabel('Order date')
plt.ylabel('Total sales')
plt.show()
```

                        SALES OVER TIME
It shows the sales flctuate over time,with several high sales peaks and lower sales periods.

CUSTOMER SEGMENT

```python
segment_sales=df.groupby('Segment')['Sales'].sum()
segment_sales.plot(kind='bar',figsize=(8,5))
plt.title('Sales by customer segment')
plt.xlabel('Customer segment')
plt.ylabel('Total sales')
plt.xticks(rotation=0)
plt.show()
```

                       CUSTOMER SEGMENT
It shows which customer gets the highest sales and it shows Consumer is the highest.

TOP PRODUCTS IN THE DATASET

```python
top_products=df.groupby('Product Name')['Sales'].sum().nlargest(10)
top_products.plot(kind='pie',figsize=(10,5))
plt.title('Top 10 products')
plt.xlabel('Product name')
plt.ylabel('Total sales')
plt.xticks(rotation=90)
plt.show()
```

                             TOP 10 Prodects
In this chart shows Products and Sales and it find top 10 Prodects in the Total Sales and Canon imageCLASS 2200 Advanced Copier is the the highest product sales.

TOP 10 PROFITS





```python
top_profit=df.groupby('Product Name')['Profit'].sum().nlargest(10)
top_profit.plot(kind='pie',figsize=(8,5))
plt.title('Top 10 profits')
plt.xlabel('Total profit')
plt.ylabel('Product name')
plt.xticks(rotation=75)
plt.show()
```

                     TOP 10 PROFITS
The Total Product and Profit finding the top 10  prodect profits. Canon imageCLASS 2200 Advanced Copies generates hieghest profit.

SALES BY SUB CATEGORY

```python
sub_category_sales=df.groupby('Sub-Category')['Sales'].sum()
sub_category_sales.plot(kind='bar',figsize=(8,5))
plt.title('Sales by sub category')
plt.xlabel('Sub Category')
plt.ylabel('Sales')
plt.show()
```

                     SALES BY SUB CATEGORY
This chart shows Sub category Sales ,Phone and Chairs have the highest sales in Sub Category and the lowest category is Labels and Fasteners.

PROFIT VS DISCOUNT

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df,x='Discount',y='Profit')
plt.title('Profit vs Discount')
plt.ylabel('Discount')
plt.ylabel('Profit')
plt.show()
```

                          PROFIT VS DISCOUNT
This shows the relationship between Profit and Discount suggests that higher discounts can be associated with lower or negative profit.

SALES TREND

```python
df['Month']=df['Order Date'].dt.to_period('M')
monthly_sales=df.groupby('Month')['Sales'].sum()
monthly_sales.plot(figsize=(10,6))
plt.title('Monthly sales trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

                             MONTH AND SALES
The sales trend shows considerable variation over time, with several noticeable peaks.
